# Quantum AI Mini Lab — Experiment
Notebook singkat ini menjalankan baseline klasik dan VQC pada simulator lokal. Jalankan dari folder proyek atau sesuaikan path di sel pertama.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))

import numpy as np
from sklearn.datasets import make_moons
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler

from models.classical_model import train_classical_model, predict as classical_predict
from models.quantum_model import train_quantum_model, predict as quantum_predict, circuit_as_text
from utils.i18n import TRANSLATIONS
from utils.visualization import plot_dataset, plot_decision_boundary, plot_objective
```

```python
X, y = make_moons(n_samples=60, noise=0.16, random_state=42)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, stratify=y, random_state=42
)
scaler = MinMaxScaler(feature_range=(0, np.pi))
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
labels = TRANSLATIONS['English']
plot_dataset(np.vstack([X_train, X_test]), np.hstack([y_train, y_test]), 'make_moons', labels)
```

```python
classical, classical_seconds = train_classical_model(X_train, y_train)
classical_pred = classical_predict(classical, X_test)
print(f'Classical accuracy: {accuracy_score(y_test, classical_pred):.2%}')
print(f'Classical time:     {classical_seconds:.3f} s')
plot_decision_boundary(
    lambda values: classical_predict(classical, values),
    X_train, y_train, X_test, y_test, 'Classical decision boundary', labels
)
```

```python
print(circuit_as_text())
vqc, quantum_seconds, objective = train_quantum_model(X_train, y_train, maxiter=25, seed=42)
quantum_pred = quantum_predict(vqc, X_test)
print(f'Quantum VQC accuracy: {accuracy_score(y_test, quantum_pred):.2%}')
print(f'Quantum VQC time:     {quantum_seconds:.3f} s')
plot_objective(objective, labels)
```

## Interpretasi
Jika baseline lebih cepat atau lebih akurat, itu bukan kegagalan eksperimen. VQC di sini adalah demonstrasi bagaimana data disandikan ke circuit dan parameter circuit dilatih dalam loop hybrid quantum-classical.
