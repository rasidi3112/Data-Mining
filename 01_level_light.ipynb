## Teori: Apa Itu Data Mining?

**Data Mining** adalah proses menemukan pola, anomali, dan hubungan dalam data berskala besar menggunakan metode statistik, machine learning, dan database.

### Perbedaan dengan Statistik Tradisional

| Aspek | Statistik Tradisional | Data Mining |
|:------|:---------------------|:------------|
| Pendekatan | Hipotesis -> Uji | Data -> Pola |
| Ukuran data | Ratusan sampel | Ribuan hingga jutaan |
| Tujuan utama | Inferensi populasi | Prediksi & penemuan pola |
| Variabel | Sedikit, dipilih manual | Banyak, dieksplorasi otomatis |

### Kerangka CRISP-DM

CRISP-DM (Cross-Industry Standard Process for Data Mining) adalah metodologi standar industri:

```
┌──────────────────────┐
│  1. BUSINESS         │ -> Apa masalah bisnis yang ingin dijawab?
│     UNDERSTANDING    │   Contoh: "Mengapa pelanggan berhenti berlangganan?"
├──────────────────────┤
│  2. DATA             │ -> Bagaimana bentuk, kualitas, dan isi data?
│     UNDERSTANDING    │   <- KITA DI SINI (Level Light)
├──────────────────────┤
│  3. DATA             │ -> Pembersihan, transformasi, feature engineering
│     PREPARATION      │
├──────────────────────┤
│  4. MODELING         │ -> Membangun dan melatih model
├──────────────────────┤
│  5. EVALUATION       │ -> Mengukur performa model
├──────────────────────┤
│  6. DEPLOYMENT       │ -> Menerapkan model di lingkungan produksi
└──────────────────────┘
```

> ** Prinsip:** Jangan langsung membuat model sebelum memahami data Anda. Lebih dari 70% waktu proyek data mining dihabiskan pada tahap 1-3.

## 1. Memuat dan Mengenal Dataset

### Teori: Fitur vs Target

Dalam data mining, data terdiri dari:
- **Fitur (Features / X)**: Kolom yang digunakan untuk membuat prediksi. Disebut juga *variabel independen*, *atribut*, atau *prediktor*.
- **Target (Label / y)**: Kolom yang ingin kita prediksi. Disebut juga *variabel dependen* atau *respons*.

```
┌─────────────────────────────────────────────┐
│              Dataset                         │
│  ┌────────────────────────┐  ┌───────────┐  │
│  │   Fitur (X)            │  │ Target (y)│  │
│  │  sepal_length          │  │  species   │  │
│  │  sepal_width           │  │            │  │
│  │  petal_length          │  │            │  │
│  │  petal_width           │  │            │  │
│  └────────────────────────┘  └───────────┘  │
└─────────────────────────────────────────────┘
```

> ** Aturan Emas:** Target tidak boleh masuk ke dalam fitur! Ini menyebabkan **data leakage** (kebocoran data) yang membuat model terlihat bagus tapi sebenarnya curang.

### Dataset Iris

Dataset Iris adalah dataset klasik yang berisi pengukuran bunga Iris dari 3 spesies:
- **Setosa**: Mudah dibedakan dari yang lain
- **Versicolor**: Agak mirip dengan Virginica  
- **Virginica**: Agak mirip dengan Versicolor

Setiap sampel memiliki 4 fitur pengukuran dalam centimeter.

````python
# Memuat dataset Iris dari scikit-learn
iris = load_iris(as_frame=True)
df = iris.frame.copy()

# Menambahkan nama spesies yang mudah dibaca
df['species'] = df['target'].map(dict(enumerate(iris.target_names)))

print('='*60)
print('INFORMASI UMUM DATASET')
print('='*60)
print(f'Jumlah baris (sampel)  : {df.shape[0]}')
print(f'Jumlah kolom (variabel): {df.shape[1]}')
print(f'Nama kolom: {list(df.columns)}')
print()

# Menampilkan 5 baris pertama
print(' 5 baris pertama:')
display(df.head())
````

````python
# Info tipe data  -  penting untuk mengetahui kolom mana numerik vs kategorikal
print(' Informasi Tipe Data:')
print()
df.info()
print()
print('Penjelasan:')
print('- float64: angka desimal (numerik kontinu)')
print('- int64  : bilangan bulat (numerik diskrit / kode)')
print('- object : teks (kategorikal)')
````

## 2. Statistik Deskriptif

### Teori: Ukuran Statistik Dasar

| Ukuran | Arti | Kegunaan |
|:-------|:-----|:---------|
| **count** | Jumlah data non-null | Mendeteksi missing values |
| **mean** | Rata-rata | Pusat distribusi |
| **std** | Standar deviasi | Sebaran/variasi data |
| **min / max** | Nilai terkecil / terbesar | Rentang data, deteksi anomali |
| **25% / 50% / 75%** | Kuartil | Distribusi tanpa pengaruh outlier |

**IQR (Interquartile Range)** = Q3 - Q1 -> mengukur sebaran 50% data tengah.

````python
# Statistik deskriptif untuk kolom numerik
print(' Statistik Deskriptif (transposed agar mudah dibaca):')
desc = df.describe().T

# Tambahkan kolom IQR
desc['IQR'] = desc['75%'] - desc['25%']
desc['range'] = desc['max'] - desc['min']

display(desc.round(2))

print()
print(' Perhatikan:')
print('  - Kolom mana yang memiliki variasi (std) paling besar?')
print('  - Apakah ada kolom dengan range sangat besar yang mungkin mengandung outlier?')
````

## 3. Pemeriksaan Kualitas Data

### Teori: Mengapa Kualitas Data Penting?

Prinsip **GIGO (Garbage In, Garbage Out)**: jika data kotor, model akan menghasilkan prediksi yang buruk, tidak peduli secanggih apa algoritmanya.

Tiga masalah kualitas data yang paling umum:

| Masalah | Dampak | Cara Deteksi |
|:--------|:-------|:-------------|
| **Missing Values** | Model gagal atau bias | `df.isna().sum()` |
| **Duplikat** | Model belajar data yang sama berulang | `df.duplicated().sum()` |
| **Outlier** | Rata-rata terdistorsi, model sensitif jarak | Box plot, Z-score, IQR |

> ** Tips:** Dataset bawaan scikit-learn biasanya sudah bersih. Dataset dunia nyata (CSV dari Kaggle, database perusahaan) hampir selalu punya masalah kualitas.

````python
# === Pemeriksaan Kualitas Data Komprehensif ===

print('='*60)
print('LAPORAN KUALITAS DATA')
print('='*60)

# 1. Missing Values
quality = pd.DataFrame({
    'tipe_data': df.dtypes.astype(str),
    'non_null': df.count(),
    'missing': df.isna().sum(),
    'missing_pct': (df.isna().sum() / len(df) * 100).round(2),
    'unique': df.nunique(),
})
print('\n Ringkasan per Kolom:')
display(quality)

# 2. Duplikat
n_dup = df.duplicated().sum()
print(f'\n Jumlah baris duplikat: {n_dup}')
if n_dup > 0:
    print('  Baris duplikat:')
    display(df[df.duplicated(keep=False)])
else:
    print('   Tidak ada baris duplikat')

# 3. Distribusi kelas target
print('\n Distribusi Target (species):')
target_dist = df['species'].value_counts()
target_pct = df['species'].value_counts(normalize=True).mul(100).round(1)
target_summary = pd.DataFrame({'jumlah': target_dist, 'persentase (%)': target_pct})
display(target_summary)

print()
if target_dist.min() / target_dist.max() > 0.8:
    print('   Kelas relatif seimbang')
else:
    print('   Kelas TIDAK seimbang  -  perlu strategi khusus (akan dibahas di Level 3)')
````

## 4. Visualisasi Data (EDA)

### Teori: Jenis Visualisasi untuk EDA

| Tujuan | Tipe Visualisasi | Kapan Digunakan |
|:-------|:-----------------|:----------------|
| Distribusi 1 variabel | Histogram, KDE | Lihat sebaran dan bentuk distribusi |
| Perbandingan kategori | Box plot, Violin plot | Bandingkan distribusi antar kelompok |
| Hubungan 2 variabel | Scatter plot | Deteksi korelasi dan pola |
| Korelasi banyak variabel | Heatmap | Identifikasi fitur yang berkorelasi kuat |
| Distribusi target | Bar chart | Cek keseimbangan kelas |

````python
# 4.1 Histogram: distribusi setiap fitur
# ───────────────────────────────────────
fig, axes = plt.subplots(2, 2, figsize=(12, 9))
fig.suptitle('Distribusi Setiap Fitur per Spesies', fontsize=14, fontweight='bold')

numeric_cols = ['sepal length (cm)', 'sepal width (cm)', 
                'petal length (cm)', 'petal width (cm)']

for ax, col in zip(axes.flat, numeric_cols):
    for species in iris.target_names:
        subset = df[df['species'] == species][col]
        ax.hist(subset, bins=15, alpha=0.6, label=species, edgecolor='white')
    ax.set_xlabel(col)
    ax.set_ylabel('Frekuensi')
    ax.legend(fontsize=8)

plt.tight_layout()
plt.show()

print(' Perhatikan: Fitur mana yang paling baik memisahkan ketiga spesies?')
````

````python
# 4.2 Box Plot: deteksi outlier dan bandingkan distribusi antar spesies
# ─────────────────────────────────────────────────────────────────────
fig, axes = plt.subplots(1, 4, figsize=(16, 5))
fig.suptitle('Box Plot: Distribusi Fitur per Spesies', fontsize=14, fontweight='bold')

for ax, col in zip(axes.flat, numeric_cols):
    sns.boxplot(data=df, x='species', y=col, ax=ax, palette='Set2')
    ax.set_xlabel('')
    ax.tick_params(axis='x', rotation=30)

plt.tight_layout()
plt.show()

print(' Titik-titik di luar kumis (whisker) box plot menunjukkan potensi outlier.')
print('   Pada dataset ini, beberapa outlier terlihat di sepal width spesies virginica.')
````

````python
# 4.3 Scatter Plot: hubungan dua fitur
# ─────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Scatter 1: petal length vs petal width
for species in iris.target_names:
    mask = df['species'] == species
    axes[0].scatter(df.loc[mask, 'petal length (cm)'], 
                    df.loc[mask, 'petal width (cm)'], 
                    label=species, alpha=0.7, edgecolors='w', s=60)
axes[0].set_xlabel('Panjang petal (cm)')
axes[0].set_ylabel('Lebar petal (cm)')
axes[0].set_title('Petal: Pemisahan Spesies Sangat Baik')
axes[0].legend(title='Spesies')

# Scatter 2: sepal length vs sepal width
for species in iris.target_names:
    mask = df['species'] == species
    axes[1].scatter(df.loc[mask, 'sepal length (cm)'], 
                    df.loc[mask, 'sepal width (cm)'], 
                    label=species, alpha=0.7, edgecolors='w', s=60)
axes[1].set_xlabel('Panjang sepal (cm)')
axes[1].set_ylabel('Lebar sepal (cm)')
axes[1].set_title('Sepal: Pemisahan Kurang Jelas')
axes[1].legend(title='Spesies')

plt.tight_layout()
plt.show()

print(' Insight:')
print('  - Fitur petal memisahkan spesies jauh lebih baik daripada sepal')
print('  - Setosa sangat mudah dipisahkan (cluster terpisah)')
print('  - Versicolor dan Virginica memiliki overlap di area petal')
````

````python
# 4.4 Correlation Heatmap: deteksi hubungan antar fitur
# ──────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(8, 6))
corr = df[numeric_cols].corr()

sns.heatmap(corr, annot=True, fmt='.2f', cmap='RdYlBu_r', center=0,
            square=True, linewidths=1, ax=ax,
            vmin=-1, vmax=1)
ax.set_title('Korelasi Antar Fitur Numerik', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()

print(' Cara membaca korelasi:')
print('  - Nilai mendekati +1: hubungan positif kuat (satu naik, lainnya naik)')
print('  - Nilai mendekati -1: hubungan negatif kuat (satu naik, lainnya turun)')
print('  - Nilai mendekati  0: tidak ada hubungan linier')
print()
print('   petal_length dan petal_width sangat berkorelasi (>0.96).')
print('    Ini berarti salah satu bisa jadi redundan (akan dibahas di level lanjut).')
````

````python
# 4.5 Pair Plot: semua kombinasi fitur sekaligus (BONUS)
# ──────────────────────────────────────────────────────
print(' Pair plot  -  mungkin butuh beberapa detik...')
g = sns.pairplot(df[numeric_cols + ['species']], hue='species', 
                 diag_kind='kde', plot_kws={'alpha': 0.6, 'edgecolor': 'w'},
                 height=2.5)
g.fig.suptitle('Pair Plot: Semua Kombinasi Fitur', y=1.02, fontsize=14, fontweight='bold')
plt.show()

print(' Pair plot memberikan gambaran menyeluruh tentang semua hubungan antar fitur.')
````

## 5. Deteksi Outlier dengan IQR

### Teori: Metode IQR

Outlier adalah data yang jauh berbeda dari mayoritas. Metode IQR mendefinisikan outlier sebagai:

```
Batas bawah = Q1 - 1.5 × IQR
Batas atas   = Q3 + 1.5 × IQR

di mana IQR = Q3 - Q1
```

Data di luar batas ini dianggap outlier.

````python
# Deteksi outlier menggunakan metode IQR
print('='*60)
print('DETEKSI OUTLIER (Metode IQR)')
print('='*60)

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    n_outliers = len(outliers)
    
    status = '' if n_outliers > 0 else ''
    print(f'{status} {col}: {n_outliers} outlier '
          f'(batas: [{lower:.2f}, {upper:.2f}])')

print()
print(' Pada dataset Iris, outlier yang terdeteksi umumnya adalah')
print('   titik data yang sah secara biologis, bukan kesalahan pengukuran.')
print('   Keputusan menghapus/menyimpan outlier harus berdasarkan konteks domain.')
````
