## Metodologi CRISP-DM

Seluruh modul menerapkan kerangka kerja standar CRISP-DM (Cross-Industry Standard Process for Data Mining):
1. Business Understanding: Menentukan masalah dan tujuan analitik.
2. Data Understanding: Eksplorasi karakteristik dan kualitas data.
3. Data Preparation: Pembersihan, transformasi, dan preprocessing fitur.
4. Modeling: Membangun serta melatih algoritma model.
5. Evaluation: Mengukur performa model dengan metrik yang tepat.
6. Deployment: Menyiapkan model untuk pemakaian nyata.

Catatan: Kualitas pemahaman data dan pencegahan data leakage jauh lebih krusial daripada sekadar mengejar angka akurasi tinggi.
